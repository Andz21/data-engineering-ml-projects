How to add kernels to your jupter notebooks within AzureML Studio

1. Open the terminal and create a new conda environment. (You can specify a Python version, but it’s optional.)



```python
conda create -n myenv python=3.13
```

2. Activate the conda environment

```python
conda activate myenv
```

3. Install the necessary packages that you require using pip install <package_name> or using pip install -r requirements.txt if you have a text file.

4. To set the newly created venev as a kernel follow the below steps

```python
conda install ipykernel
```

5. Register your environment as a kernel

```python
python -m ipykernel install --user --name myenv --display-name "Python myenv-kernel"
```

6. Open your jupter notebook and you should be able to select the newly created kernel

7. You can easily install additional packages to that specific environment by activating it first in the terminal, then using either pip or conda

```python
conda activate ENV-NAME
pip install PACKAGE-NAME
```

8. or you can make use of the magic functions using % in your jupyter script to execute command in your selected kernel

```python
%pip install PACKAGE-NAME
```

9. If you ever need to deactivate the environment and go back to the base environment, simply use:

```python
conda deactivate
```

10. And if you find that you no longer need the environment, you can remove it with:

```python
conda env remove --name ENV-NAME
```

11. However, remember that if you remove the environment, you will also need to remove the kernel associated with it. You can list the kernels currently installed by using:    

```python
jupyter kernelspec list
```

and then uninstall it by using:

```python
jupyter kernelspec uninstall ENV-NAME
```
