```python
from azure.identity import EnvironmentCredential,get_bearer_token_provider
import requests
import os
from dotenv import dotenv_values
from openai import AzureOpenAI
from azure.search.documents.indexes import SearchIndexClient
from azure.search.documents import SearchClient
```

```python
env_config_file = ".env"
search_endpoint = "https://azsearch-tutorial-001.search.windows.net"
index_name = "test-index"
```

```python
if env_config_file != 'None':
    if os.path.isfile(env_config_file):
        config = dotenv_values(env_config_file)
        os.environ["AZURE_TENANT_ID"] = config["AZURE_TENANT_ID"]
        os.environ["AZURE_CLIENT_ID"] = config["AZURE_CLIENT_ID"]
        os.environ["AZURE_CLIENT_SECRET"] = config["AZURE_CLIENT_SECRET"]
        credential = EnvironmentCredential()
        
print(credential)
```

```python
# To check if azure search is accessible from vs code

index_client = SearchIndexClient(endpoint=search_endpoint,credential=credential)
search_client = SearchClient(endpoint=search_endpoint,credential=credential,index_name=index_name)

index_list = index_client.list_index_names()

if index_name not in index_list:
    print(f"The index '{index_name}' does not exist. Please create it.")
else:
    print("Index exists.")
```
